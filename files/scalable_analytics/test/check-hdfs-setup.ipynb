# Checking HDFS setup

Launch Jupyter from `nix develop .#spark-hadoop` and select **Python (Spark/Hadoop)**. The cells use `%%bash` to run Hadoop commands. `rsm-hdfs init` preserves existing data; use `rsm-hdfs reset --yes` only when you intend to erase it.


```python
%%bash
# Hadoop installation and active filesystem
echo "$HADOOP_HOME"
hdfs getconf -confKey fs.defaultFS
```

```python
%%bash
# Initialize once; rsm-hdfs preserves existing HDFS data
if command -v rsm-hdfs >/dev/null; then rsm-hdfs init; else /opt/hadoop/init-dfs.sh; fi
```

```python
%%bash
# start the HDFS services
if command -v rsm-hdfs >/dev/null; then rsm-hdfs start; else /opt/hadoop/start-dfs.sh; fi
```

```python
%%bash
# check Java processes running
jps
```

```python
%%bash
# check status of name node and data node
if command -v rsm-hdfs >/dev/null; then rsm-hdfs urls; else echo http://localhost:9870/; fi
```

```python
%%bash
# check status of data node
if command -v rsm-hdfs >/dev/null; then rsm-hdfs urls; else echo http://localhost:9864/; fi
```

```python
%%bash
# check status of secondary name node
if command -v rsm-hdfs >/dev/null; then rsm-hdfs urls; else echo http://localhost:9868/; fi
```

```python
%%bash
# stop the HDFS services
if command -v rsm-hdfs >/dev/null; then rsm-hdfs stop; else /opt/hadoop/stop-dfs.sh; fi
```

```python
%%bash
# check if any Java processes are still running
jps
```
